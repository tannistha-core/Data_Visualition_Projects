```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
def basic_exploration(df):
    print("First 5 rows:")
    display(df.head())

    print("\nShape of dataset:")
    print(df.shape)

    print("\nDataset information:")
    df.info()

    print("\nStatistical summary:")
    display(df.describe(include="all"))

    print("\nMissing values:")
    display(df.isnull().sum())
```

```python
iris = sns.load_dataset("iris")
basic_exploration(iris)
```

## Find the count of the species

```python
print("Flower species:")
print(iris["species"].unique())

sns.countplot(data=iris, x="species")
plt.title("Count of Iris Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```

## Find Average Sepal Length and Petal Length

```python
print("Average sepal length:", iris["sepal_length"].mean())
print("Average petal length:", iris["petal_length"].mean())

print("\nAverage measurements by species:")
display(iris.groupby("species").mean(numeric_only=True))
```

## Compare Petal Length of Different Species

```python
sns.boxplot(data=iris, x="species", y="petal_length")
plt.title("Petal Length by Iris Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")
plt.show()
```

## Observation
Setosa has the smallest petal length, while Virginica has the largest petal length. Petal length is a strong feature for separating species.

## Compare Sepal Width of Different Species

```python
sns.violinplot(data=iris, x="species", y="sepal_width")
plt.title("Sepal Width by Iris Species")
plt.xlabel("Species")
plt.ylabel("Sepal Width")
plt.show()
```

## Observation
Sepal width shows some difference between species, but it is not as strong as petal length for classification

## Create Correlation Heatmap

```python
corr = iris.drop(columns=["species"]).corr()

sns.heatmap(corr, annot=True, cmap="RdPu")
plt.title("Correlation Heatmap of Iris Features")
plt.show()
```

## Observation
Petal length and petal width are strongly positively correlated. Sepal length also has a positive relation with petal length and petal width.

## Overall comparison

```python
sns.pairplot(iris, hue="species", palette="husl")
plt.show()
```
